## Deploying and Monitoring Agents on Databricks

This course covers the end-to-end lifecycle for deploying and monitoring generative AI agents on Databricks. Participants will learn how to deploy agents as Databricks Apps using Declarative Automation Bundles (DABs), integrate tools via the Model Context Protocol (MCP), instrument agents with MLflow Tracing, and evaluate production quality using scorers, multi-turn judges, and online evaluation. Through hands-on demos and labs, participants will gain practical experience building, observing, and monitoring production-grade AI agents on the Databricks platform.

----

## Prerequisites
The content was developed for participants with these skills/knowledge/abilities:
- Familiarity with Databricks workspace and notebooks
- Familiarity with Unity Catalog
- Experience building agents using the OpenAI Agents SDK
- Basic knowledge of MLflow and Python
- Familiarity with GenAI agent concepts (LLM calls, tool invocation, retrieval)

## Course Agenda
The following modules are part of the **Generative AI Application Deployment and Monitoring** course by **Databricks Academy**.

| # | Type | Notebook | Description |
|---|------|----------|-------------|
| 1 | Lecture | [Agent Deployment on Databricks]($./1 - Lecture - Agent Deployment on Databricks) | Agent lifecycle, Databricks Apps, DABs, deployment patterns, resource injection |
| 2 | Demo | [Deploying an Observable Agent]($./2 - Demo - Deploying an Observable Agent) | Progressively build an agent: deploy as an App, add MLflow tracing, integrate MCP tools |
| 3 | Lecture | [Tool Integration and Observability]($./3 - Lecture - Tool Integration and Observability) | Model Context Protocol, MCP server types, MCP lifecycle, MLflow Tracing foundations |
| 4 | Demo | [Tracing for Production Agents]($./4 - Demo - Tracing for Production Agents) | Manual spans, custom attributes, multi-turn session grouping, programmatic trace search |
| 5 | Lecture | [Production Evaluation and Monitoring]($./5 - Lecture - Production Evaluation and Monitoring) | Scorer types, custom scorers, multi-turn judges, sampling strategies, scorer lifecycle |
| 6 | Lab | [Online Evaluation with Multi-Turn Conversations]($./6 - Lab - Online Evaluation with Multi-Turn Conversations) | Deploy judges, create custom scorers, tune sampling rates, run the metrics computation job |
| 7 | Lab | [Backfilling and Archiving Multi-Turn Conversations]($./7 - Lab - Backfilling and Archiving Multi-Turn Conversations) | Backfill historical traces, enable/disable trace archival to Delta, scorer cleanup |
