## Building Agentic Applications on Databricks

This course teaches students how to build production-grade agentic applications on Databricks. Students learn to create governed agent tools using Unity Catalog and MCP, build single and multi-agent systems with the OpenAI Agents SDK, and build agents with other popular frameworks such as LangChain and DSPy. The course covers the full progression from tool prototyping to production deployment, with hands-on experience using MLflow tracing to observe agent execution. Through practical demonstrations and labs, students develop skills in building, orchestrating, and governing AI agents on the Databricks platform.

## Prerequisites

### Python-Specific Skills
- Basic Python syntax and data structures (lists, dictionaries)
- Understanding of functions, classes, and decorators
- Experience with Python package management and imports
- Familiarity with JSON data handling
- Basic understanding of async/await patterns

### SQL-Specific Skills
- Basic SQL query syntax (SELECT, FROM, WHERE)
- Understanding of SQL functions and user-defined functions
- Experience with Unity Catalog SQL functions

### Databricks-Specific Skills
- Understanding of Databricks workspace navigation and notebook interface
- Knowledge of Unity Catalog structure (catalogs, schemas, tables, volumes, functions)
- Experience with Databricks compute resources and serverless computing
- Familiarity with MLflow experiment tracking
- Understanding of Databricks model serving endpoints

### GenAI/Agent-Specific Skills
- Basic understanding of Large Language Models (LLMs) and their capabilities
- Knowledge of prompt engineering and system prompts
- Familiarity with tool-calling agents and function calling concepts
- Basic awareness of the Model Context Protocol (MCP)

### Other/Optional Skills
- Experience with YAML configuration files can be helpful
- Familiarity with LangChain or the OpenAI SDK can be helpful
- Basic understanding of REST APIs and HTTP requests can be helpful

## Course Learning Objectives

By the end of this course, you will be able to:

1. **Define AI agents and their architectures** — Explain what AI agents are, identify their core components (reasoning, planning, tool use), and determine when to use a single-agent vs. multi-agent architecture.
2. **Create governed agent tools in Unity Catalog** — Build and register both SQL and Python functions in Unity Catalog, equipping them with descriptive names, type hints, and structured docstrings that enable reliable agent tool selection.
3. **Assess and improve tool documentation for agent use** — Identify documentation gaps that cause agents to misuse tools, and rewrite function signatures and docstrings to improve agent behavior.
4. **Expose and discover tools via the Model Context Protocol (MCP)** — Use Databricks Managed MCP servers to connect agents to Unity Catalog functions, AI Search indexes, and other Databricks services at runtime.
5. **Build single-agent systems with the OpenAI Agents SDK** — Configure the SDK with Databricks model routing, define custom tools with `@function_tool`, and build a functional agent that reasons over and calls tools.
6. **Orchestrate multi-agent systems with handoffs** — Build supervisor-worker architectures using the OpenAI Agents SDK, route requests to specialized worker agents via handoffs, and refine routing through YAML-based instructions.
7. **Build agents with LangChain and DSPy on Databricks** — Apply the same tool-creation-to-agent workflow using LangChain's `@tool` decorator and `create_agent`, as well as DSPy signatures and `dspy.ReAct` modules.
8. **Observe agent execution with MLflow tracing** — Enable MLflow autologging, inspect hierarchical span traces across single- and multi-agent flows, and use tracing to debug and improve agent quality.
9. **Apply Unity Gateway governance to AI agents** — Understand how Unity Gateway provides centralized access control, rate limiting, spend tracking, and fallback routing for model services across agent applications.

## Requirements

Please review the following requirements before starting the lesson:

* To run demo and lab notebooks, you need to use **Serverless compute (version 6)**, which is enabled by default.
* This course requires access to Unity Catalog for tool registration and governance.
* The demos and labs call Databricks-hosted foundation models through Unity Gateway.

The course is structured into three logical sections:

1. **Foundations (01-03)** introduces agents, MCP, and AI governance on Databricks, then walks through building and assessing agent tools registered to Unity Catalog
2. **Building Agents with the OpenAI Agents SDK (04-06)** surveys popular agent frameworks and rapid tool prototyping, builds a single agent with MCP tools and MLflow tracing, then has you build a multi-agent system in a hands-on lab
3. **Other Frameworks (07-08)** applies the same patterns with LangChain, then has you build a tool-calling agent with DSPy in a hands-on lab

Each section combines lectures for conceptual understanding with hands-on demos and labs for practical implementation. The progression moves from tool creation to single agents to multi-agent orchestration, and then to other frameworks, ensuring students build a comprehensive understanding of agentic application development on Databricks.
